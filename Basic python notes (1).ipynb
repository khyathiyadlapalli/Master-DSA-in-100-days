Write a Python function that accepts two integer numbers. If the product of the two numbers is less than or equal to 1000, return their product; otherwise, return their sum.

```python
num1=int(input()) #Using if and else clause
num2=int(input())
b=num1*num2
c=num1+num2
if b<=1000:
    print(b)
else:
    print(c)
```

9.Write a program to count the total number of vowels (a, e, i, o, u) present in a given sentence.

```python
string=input()
vowels=['a','e','i','o','u']
count=0
for i in string:
    
    if i in vowels:
        count+=1
    
print(count)
```

15.Given an integer array nums, return true if any value appears more than once in the array, otherwise return false.

```python
nums=list(map(int,input().split(','))) #Most optimised approach and will execute for large input
seen=set()
for num in nums:
    if num in seen:
        print(True)
    else:
        seen.add(num)
print(False)
```

```python
nums=list(map(int,input().split(',')))
c=[]
for i in range(len(nums)):
    if nums[i] not in c:
        print(True)
        break
    c.append(nums[i])
else:
    print(False)
        
```
